# Apartment Maintainence Analysis 

Project Overview

This project uses Python and Monte Carlo simulation to analyze the weekly maintenance workload of a 500-apartment complex. It estimates the number of maintenance requests, the labor hours required, and the associated staffing costs.

The analysis also compares different staffing levels to identify the number of employees that minimizes the estimated weekly operating cost.

```python
import numpy as np

rng = np.random.default_rng(42)
```

Set up the parameters

```python
n_apartments = 500
request_probability = 0.15
n_staff = 3
n_trials = 1000

staff_capacity = n_staff * 40
staff_cost = n_staff * 1000
overtime_rate = 50
```

An empty list to store each week's cost

```python
weekly_costs = []
```

5) Simulate 1000 trials

```python
rng = np.random.default_rng(42)
n_trials = 1000
weekly_costs = []
for trial in range(n_trials):
    number_of_requests = rng.binomial(n_apartments, request_probability)
    request_hours = rng.uniform(1, 3, size=number_of_requests)
    total_hours = request_hours.sum()

    overtime_hours = max(0, total_hours - staff_capacity)
    weekly_cost = staff_cost + overtime_hours * overtime_rate

    weekly_costs.append(weekly_cost)
```

```python
# Average the costs across all 1,000 simulated weeks
expected_weekly_cost = np.mean(weekly_costs)
print(f"Estimated expected weekly cost: ${expected_weekly_cost:,.2f}")
```

6) Standard deviation of weekly cost

```python
std_weekly_cost = np.std(weekly_costs, ddof=1)

print(f"Sample standard deviation: ${std_weekly_cost:,.2f}")
```

7 & 8) Confidence limit

```python
sample_mean = np.mean(weekly_costs)
sample_std = np.std(weekly_costs, ddof=1)
sample_size = len(weekly_costs)

# Calculate uncertainty in the estimated mean
standard_error = sample_std / np.sqrt(sample_size)
margin_of_error = 1.96 * standard_error

# Calculate the lower and upper confidence limits
lower_limit = sample_mean - margin_of_error
upper_limit = sample_mean + margin_of_error

print(f"Lower confidence limit: ${lower_limit:,.2f}")
print(f"Upper confidence limit: ${upper_limit:,.2f}")
```

9 & 10) Staffing level and minimum estimated mean cost (set the seed again)

```python
import numpy as np
rng = np.random.default_rng(42)

n_apartments = 500
request_probability = 0.15
n_trials = 1000

# Store the average cost for each staffing level
mean_costs_by_staff = []

# Try staffing levels 1 through 10
for staff in range(1, 11):
    capacity = staff * 40
    regular_cost = staff * 1000
    costs_for_this_staff = []

    # Simulate 1,000 weeks for this staffing level
    for trial in range(n_trials):
        number_of_requests = rng.binomial(
            n_apartments, request_probability
        )

        request_hours = rng.uniform(
            1, 3, size=number_of_requests
        )

        total_hours = request_hours.sum()
        overtime_hours = max(0, total_hours - capacity)
        weekly_cost = regular_cost + overtime_hours * 50

        costs_for_this_staff.append(weekly_cost)

    # Average the 1,000 costs for this staffing level
    mean_cost = np.mean(costs_for_this_staff)
    mean_costs_by_staff.append(mean_cost)

    print(f"{staff} staff: ${mean_cost:,.2f}")

# Find the staffing level with the lowest average cost
best_index = np.argmin(mean_costs_by_staff)
best_staff = best_index + 1
minimum_mean_cost = mean_costs_by_staff[best_index]

print(f"\nBest staffing level: {best_staff}")
print(f"Minimum estimated mean weekly cost: ${minimum_mean_cost:,.2f}")
```
